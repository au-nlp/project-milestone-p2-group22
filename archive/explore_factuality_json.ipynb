# Explore the SmolDoc factuality annotations
There are more annotations (671) than SmolDoc documents (584). How is that the case? We will look into this in this notebook.

```python
import pandas as pd
```

```python
with open("data/smoldoc-factuality-ratings.json", "r") as f:
    annotated_dataset = pd.read_json(f).T

annotated_dataset
```

```python
from utils import list_smoldoc_configs, get_smoldoc_dataset

smoldoc_configs = list_smoldoc_configs()

datasets_dict = get_smoldoc_dataset(
    configs=smoldoc_configs,
    save_path="data/smoldoc_datasets",
    force_download=False,
    verbose=False,
)

df = pd.DataFrame(datasets_dict["smoldoc__en_sw"])
df
```

```python
df.join(annotated_dataset, on="id", rsuffix="_annotated")
```

```python
missing = set(annotated_dataset.index) - set(df["id"])
print(f"There are {len(missing)} documents in the annotations that are not in the English to Swahili SmolDoc dataset.")
```

You can view which documents (their IDs) that are missing in the SmolDoc dataset by running the cell below.

```python
missing
```

Let us look at how many of the other SmolDoc documents (other language pairs) have some of the IDs not in
the English to Swahili SmolDoc document.

```python
has_missing_docs = 0

for k, dataset in datasets_dict.items():
    df = pd.DataFrame(dataset)
    if any(id_ in missing for id_ in df["id"]):
        print(f"Dataset {k} has some of the missing documents not in Swahili.")
        has_missing_docs += 1

print(f"In total, {has_missing_docs} datasets have some of the missing documents.")
```

This means that all the SmolDoc documents do not have the same source documents. So how many unique documents are there in total?

```python
all_docs_ids = set()

for _, dataset in datasets_dict.items():
    df = pd.DataFrame(dataset)
    all_docs_ids.update(set(df["id"]))

print(f"There are {len(all_docs_ids)} unique documents in all datasets combined.")
```
