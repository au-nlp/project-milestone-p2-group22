```python
import pandas as pd
```

```python
url_factuality_qa = "https://gitlab.au.dk/nlp-mnm/nlp-project/-/snippets/81/raw/main/factuality-qa.csv"
df_questions = pd.read_csv(url_factuality_qa)
df_questions = df_questions.dropna()  # Drop rows with missing QA pairs

print("Questions df shape:", df_questions.shape)
df_questions.head()
```

```python
# Dictionary mapping models and methods to their specific snippet URLs
snippet_urls = {
    "llama": {
        "sft": "https://gitlab.au.dk/nlp-mnm/nlp-project/-/snippets/84/raw/main/answers_sft_llama.csv",
        "peft": "https://gitlab.au.dk/nlp-mnm/nlp-project/-/snippets/87/raw/main/answers_peft_llama.csv"
    },
    "gemma": {
        "sft": "https://gitlab.au.dk/nlp-mnm/nlp-project/-/snippets/83/raw/main/answers_sft_gemma.csv",
        "peft": "https://gitlab.au.dk/nlp-mnm/nlp-project/-/snippets/86/raw/main/answers_peft_gemma.csv"
    },
    "deepseek": {
        "sft": "https://gitlab.au.dk/nlp-mnm/nlp-project/-/snippets/85/raw/main/answers_sft_deepseek.csv",
        "peft": "https://gitlab.au.dk/nlp-mnm/nlp-project/-/snippets/88/raw/main/answers_peft_deepseek.csv"
    }
}

# Example: Loading all 6 into a single dictionary of DataFrames
dfs = {}
for model, methods in snippet_urls.items():
    for method, url in methods.items():
        key = f"{model}_{method}"
        try:
            dfs[key] = pd.read_csv(url)
            print(f"Loaded {key}: {dfs[key].shape}")
        except Exception as e:
            print(f"Could not load {key}: {e}")

# Access a specific one:
dfs['llama_sft'].head()
```

```python
from IPython.display import display, Markdown

# Define the models and methods for the loop
models = ["llama", "gemma", "deepseek"]
methods = ["sft", "peft"]

# Set pandas to show full text without clipping
pd.set_option('display.max_colwidth', None)

for model in models:
    for method in methods:
        key = f"{model}_{method}"

        # 1. Create a clear, bold header for GitHub readers
        display(Markdown(f"---"))
        display(Markdown(f"## 🤖 Model: {model.upper()} | Method: {method.upper()}"))

        # 2. Select the questions and the corresponding model answers
        # We merge them so the question is right next to the answer
        sample_display = df_questions[['question']].head(5).copy()
        sample_display['Model Answer'] = dfs[key]['model answer'].head(5).values

        # 3. Apply basic styling that renders well on GitHub
        styled_output = sample_display.style.set_properties(**{
            'text-align': 'left',
            'white-space': 'pre-wrap' # Keeps line breaks in the answers
        }).set_table_styles([
            {'selector': 'th', 'props': [('background-color', '#f2f2f2'), ('color', '#333')]}
        ])

        display(styled_output)
```
